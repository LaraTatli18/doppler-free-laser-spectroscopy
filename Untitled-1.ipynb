```python
from header import *

print("\nCalibrating:")


#group the data
_,data=load("Data\data_broad_nopump\\")


GROUP_SIZE=10
N_GROUPS=N_POINTS//GROUP_SIZE

grouped_data=group(data, N_GROUPS, GROUP_SIZE)

#truncate the calibration data
i=18000
f=75000
truncated_grouped_data=grouped_data[:,i//GROUP_SIZE:f//GROUP_SIZE]

print("Fitting Curves")
def fit(t,A,A_1,phi_0,phi_1,phi_2,C0,C1,C2,t_0,t_1):
    #y=A*(np.sin(phi_2*t**2+phi_1*t+phi_0))+(C2*t**2+C1*t+C0)
    y=(A+A_1*t)*np.sin(phi_0+phi_1*np.sqrt((t-t_0)**2+phi_2))+C0+C1*np.sqrt((t-t_1)**2+C2**2)
    return y

def fite(t,A,A_1,phi_0,phi_1,phi_2,C0,C1,C2, t_0, t_1):
    return (A+A_1*t)*unp.sin(phi_0+phi_1*unp.sqrt((t-t_0)**2+phi_2))+C0+C1*unp.sqrt((t-t_1)**2+C2**2)

p0=[5.2, 0, 5.6, 0.00044, 1000000, 15, -0.00038, 10000,10000,10000]

paramiters=uarray(np.zeros((N_RUNS, len(p0))),np.zeros((N_RUNS, len(p0))))

for run in range(10):
    popt,pcov=curve_fit(
        f=fit,
        xdata=np.arange(i,f,GROUP_SIZE),
        ydata=unp.nominal_values(truncated_grouped_data[run]),
        p0=p0,
        sigma=unp.std_devs(truncated_grouped_data[run]),
        absolute_sigma=True,
        maxfev=10000000
    )
    paramiters[run]=uncrt.correlated_values(popt,pcov)

    #check chi squared
    reduced_chi_squared=np.sum((fite(np.arange(i,f,GROUP_SIZE),*paramiters[run])-truncated_grouped_data[run])**2/unp.std_devs(truncated_grouped_data[run])**2)/(len(np.arange(i,f,GROUP_SIZE))-len(p0))

    if(reduced_chi_squared>2):
        print("Warning high reduced chi_squared value:",reduced_chi_squared)
    if(reduced_chi_squared<0.5):
        print("Warning low reduced chi_squared value:",reduced_chi_squared)


print("Creating Function")

L1=ufloat(0.075,0.003)
L2=ufloat(0.37,0.002)


p2=ufloat(0,0)
p3=ufloat(0,0)

for j in range(N_RUNS):
    p2+=paramiters[j,2]
    p3+=paramiters[j,3]

p2=p2/N_RUNS
p3=p3/N_RUNS

def d(t,i,paramiters=paramiters):
    #return constants.c*(p2*t+p3*t**2)/(4*np.pi*(L1-L2))*10**(-9)
    phi_1=paramiters[i,3]
    phi_2=paramiters[i,4]
    t_0=paramiters[i,8]
    return phi_1*unp.sqrt((t-t_0)**2+phi_2)


print("  Calibration Complete\n")

```

```python
print("Fitting Curves")
def fit2(t,A,phi_0,phi_1,phi_2,C0,C1,C2):
    y=A*(np.sin(phi_2*t**2+phi_1*t+phi_0))+(C2*t**2+C1*t+C0)
    return y

def fite2(t,A,phi_0,phi_1,phi_2,C0,C1,C2):
    return A*(unp.sin(phi_2*t**2+phi_1*t+phi_0))+(C2*t**2+C1*t+C0)

p02=[5.2, 0, 0.00044, 0, 15, 0,0]

paramiters2=uarray(np.zeros((N_RUNS, len(p02))),np.zeros((N_RUNS, len(p02))))

for run in range(N_RUNS):
    popt,pcov=curve_fit(
        f=fit2,
        xdata=np.arange(i,f,GROUP_SIZE),
        ydata=unp.nominal_values(truncated_grouped_data[run]),
        p0=p02,
        sigma=unp.std_devs(truncated_grouped_data[run]),
        absolute_sigma=True,
        maxfev=10000000
    )
    paramiters2[run]=uncrt.correlated_values(popt,pcov)

    #check chi squared
    reduced_chi_squared=np.sum((fite2(np.arange(i,f,GROUP_SIZE),*paramiters2[run])-truncated_grouped_data[run])**2/unp.std_devs(truncated_grouped_data[run])**2)/(len(np.arange(i,f,GROUP_SIZE))-len(p02))

    if(reduced_chi_squared>2):
        print("Warning high reduced chi_squared value:",reduced_chi_squared)
    if(reduced_chi_squared<0.5):
        print("Warning low reduced chi_squared value:",reduced_chi_squared)
```

```python
print("Fitting Curves")
def fit4(t,A_0,A_1,phi_0,phi_1,phi_2,phi_3,C0,C1,C2,C3):
    y=(A_0+A_1*t)*(np.sin(phi_3*t**3+phi_2*t**2+phi_1*t+phi_0))+(C3*t**3*C2*t**2+C1*t+C0)
    return y

def fite4(t,A_0,A_1,phi_0,phi_1,phi_2, phi_3, C0,C1,C2,C3):
    return (A_0+A_1*t)*(unp.sin(phi_3*t**3+phi_2*t**2+phi_1*t+phi_0))+(C3*t**3*C2*t**2+C1*t+C0)

p04=[5.2,0, 0, 0.00044,0,0, 15, 0,0,0]

paramiters4=uarray(np.zeros((N_RUNS, len(p04))),np.zeros((N_RUNS, len(p04))))

for run in range(N_RUNS):
    popt,pcov=curve_fit(
        f=fit4,
        xdata=np.arange(i,f,GROUP_SIZE),
        ydata=unp.nominal_values(truncated_grouped_data[run]),
        p0=p04,
        sigma=unp.std_devs(truncated_grouped_data[run]),
        absolute_sigma=True,
        maxfev=1000000000
    )
    paramiters4[run]=uncrt.correlated_values(popt,pcov)

    #check chi squared
    reduced_chi_squared=np.sum((fite4(np.arange(i,f,GROUP_SIZE),*paramiters4[run])-truncated_grouped_data[run])**2/unp.std_devs(truncated_grouped_data[run])**2)/(len(np.arange(i,f,GROUP_SIZE))-len(p04))

    if(reduced_chi_squared>2):
        print("Warning high reduced chi_squared value:",reduced_chi_squared)
    if(reduced_chi_squared<0.5):
        print("Warning low reduced chi_squared value:",reduced_chi_squared)
```
